```python

%%writefile icici.py

import streamlit as st
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score


# Page configuration
st.set_page_config(
    page_title="ICICI Bank Stock Price Prediction",
    page_icon="📈",
    layout="wide"
)

st.title("ICICI Bank Stock Price Prediction")
st.write("Machine Learning Based Stock Price Analysis")


# 1. Load Dataset
file_path = "/content/ICICIBANK.NS.csv"

try:
    df = pd.read_csv(file_path)
except Exception as e:
    st.error("Unable to load the ICICI Bank dataset.")
    st.write(str(e))
    st.stop()


df.columns = df.columns.str.strip()

required_columns = ["Open", "High", "Low", "Close", "Volume"]

missing_columns = [
    col for col in required_columns if col not in df.columns
]

if missing_columns:
    st.error(f"Missing columns: {missing_columns}")
    st.stop()


# 2. Data Preprocessing
if "Date" in df.columns:
    df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
    df = df.sort_values("Date")

for col in required_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")

df.replace([np.inf, -np.inf], np.nan, inplace=True)
df.dropna(subset=required_columns, inplace=True)

if len(df) < 10:
    st.error("Insufficient valid data for model training.")
    st.stop()

st.success("ICICI Bank dataset loaded successfully!")


# 3. Dataset Overview
st.header("Dataset Overview")

total_records = len(df)
opening_price = df["Open"].iloc[-1]
closing_price = df["Close"].iloc[-1]

col1, col2, col3 = st.columns(3)

with col1:
    st.write("**Total Records**")
    st.subheader(f"{total_records:,}")

with col2:
    st.write("**Opening Price**")
    st.subheader(f"₹{opening_price:.2f}")

with col3:
    st.write("**Closing Price**")
    st.subheader(f"₹{closing_price:.2f}")


# 4. Price Change Analysis
df["Price Change"] = df["Close"] - df["Open"]

st.header("Price Change Distribution")

fig, ax = plt.subplots(figsize=(9, 4))

sns.histplot(
    df["Price Change"],
    bins=50,
    kde=True,
    ax=ax
)

ax.set_title("Distribution of Price Change")
ax.set_xlabel("Price Change")
ax.set_ylabel("Frequency")

st.pyplot(fig)
plt.close(fig)


# 5. Feature Selection
features = ["Open", "High", "Low", "Volume"]
target = "Close"

X = df[features]
y = df[target]


# 6. Train-Test Split
split_index = int(len(df) * 0.8)

X_train = X.iloc[:split_index]
X_test = X.iloc[split_index:]

y_train = y.iloc[:split_index]
y_test = y.iloc[split_index:]


# 7. Random Forest Model
model = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

model.fit(X_train, y_train)


# 8. Prediction
y_pred = model.predict(X_test)


# 9. Model Performance
mse = mean_squared_error(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

st.header("Model Performance")

col1, col2, col3 = st.columns(3)

with col1:
    st.write("**Mean Squared Error (MSE)**")
    st.subheader(f"{mse:.4f}")

with col2:
    st.write("**Mean Absolute Error (MAE)**")
    st.subheader(f"{mae:.4f}")

with col3:
    st.write("**R² Score**")
    st.subheader(f"{r2:.4f}")


# 10. Feature Importance
st.header("Feature Importance")

importance_df = pd.DataFrame({
    "Feature": features,
    "Importance": model.feature_importances_
})

importance_df = importance_df.sort_values(
    by="Importance",
    ascending=False
)

fig, ax = plt.subplots(figsize=(9, 4))

sns.barplot(
    data=importance_df,
    x="Importance",
    y="Feature",
    ax=ax
)

ax.set_title("Random Forest Feature Importance")
ax.set_xlabel("Importance")
ax.set_ylabel("Features")

st.pyplot(fig)
plt.close(fig)


# 11. Actual vs Predicted Prices
st.header("Actual vs Predicted Stock Prices")

fig, ax = plt.subplots(figsize=(8, 4))

ax.scatter(y_test, y_pred, alpha=0.6)

min_price = min(y_test.min(), y_pred.min())
max_price = max(y_test.max(), y_pred.max())

ax.plot(
    [min_price, max_price],
    [min_price, max_price],
    linestyle="--"
)

ax.set_xlabel("Actual Close Price")
ax.set_ylabel("Predicted Close Price")
ax.set_title("Actual vs Predicted Prices")

st.pyplot(fig)
plt.close(fig)


# 12. Prediction Results
st.header("Prediction Results")

results = pd.DataFrame({
    "Actual Close Price": y_test.values,
    "Predicted Close Price": y_pred
})

# Display results without st.dataframe()
st.text(results.to_string(
    index=False,
    float_format=lambda x: f"{x:.2f}"
))


# 13. Final Status
st.success("Stock Price Prediction Completed Successfully!")
st.write("Model: Random Forest Regressor")
st.write("Training Data: 80%")
st.write("Testing Data: 20%")
```
